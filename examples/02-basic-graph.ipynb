# Styles, layouts, and grouping

This notebook covers the three main ways to enrich a graph visualization:
binding visual properties to data, running automatic layouts, and collapsing
nodes into groups.


```python
import ogma_jupyter as og
from ogma_jupyter import rules

og.set_license("your-license-key")  # or set OGMA_LICENSE_KEY env var
```

## Sample data

A small org-chart graph. Each person has a `role`, `department`, and `score`.


```python
GRAPH = {
    "nodes": [
        {"id": "alice",   "data": {"name": "Alice",   "role": "engineer",  "department": "eng",     "score": 80}},
        {"id": "bob",     "data": {"name": "Bob",     "role": "manager",   "department": "eng",     "score": 55}},
        {"id": "carol",   "data": {"name": "Carol",   "role": "designer",  "department": "design",  "score": 70}},
        {"id": "dave",    "data": {"name": "Dave",    "role": "manager",   "department": "design",  "score": 65}},
        {"id": "eve",     "data": {"name": "Eve",     "role": "engineer",  "department": "eng",     "score": 90}},
    ],
    "edges": [
        {"source": "alice", "target": "bob"},
        {"source": "eve",   "target": "bob"},
        {"source": "carol", "target": "dave"},
        {"source": "bob",   "target": "dave"},
    ],
}
```

## 1. Style rules

`add_style_rule()` binds visual attributes to node or edge data.
Use `rules.map()` for categorical fields, `rules.slices()` for numerical ranges,
and `rules.template()` for text labels.


```python
widget = og.OgmaWidget(graph_data=GRAPH)

# Color nodes by department
widget.add_style_rule(
    node_attributes={
        "color": rules.map(
            field="data.department",
            values={"eng": "#4e79a7", "design": "#f28e2b"},
            fallback="gray",
        ),
        # Size nodes by score (4 slices between radius 6 and 24)
        "radius": rules.slices(
            field="data.score",
            values={"nbSlices": 4, "min": 6, "max": 24},
        ),
        # Label each node with its name
        "text": {"content": rules.template("{{data.name}}")},
    }
)

# Style edges by role of source (just an example of edge rules)
widget.add_style_rule(
    edge_attributes={
        "color": "#cccccc",
        "width": 2,
    }
)

widget
```

## 2. Layouts

`run_layout()` sends a message to the JS side to re-arrange nodes.
The graph above has no explicit positions, so a layout is needed.
Run the cell below **after** the widget is displayed:


```python
widget.run_layout('force', duration=400)
```

```python
# Try hierarchical — useful for trees and org charts
widget.run_layout('hierarchical', direction='LR')
```

Available layouts: `concentric`, `force`, `forceatlas2`, `grid`,
`hierarchical`, `radial`, `sequential`.


## 3. Grouping

`group_nodes()` collapses all nodes that share a data property value
into a single meta-node. `ungroup_nodes()` restores them.


```python
widget.group_nodes(key='data.department')
```

```python
widget.ungroup_nodes()
```

## Summary

| Task | Code |
|------|------|
| Categorical color | `rules.map(field=..., values={...})` |
| Numerical size | `rules.slices(field=..., values={...})` |
| Text label | `rules.template('{{field}}')` |
| Apply rule | `widget.add_style_rule(node_attributes={...})` |
| Run layout | `widget.run_layout('force')` |
| Group nodes | `widget.group_nodes(key='data.x')` |
| Ungroup | `widget.ungroup_nodes()` |
