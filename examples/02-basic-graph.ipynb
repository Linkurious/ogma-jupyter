# Basic Graph Creation

This notebook demonstrates how to create graphs with custom node and edge attributes
using ogma-jupyter.

You'll learn:
- Creating nodes with data attributes
- Creating edges with data attributes
- Updating graph data after widget creation
- Displaying multiple widgets in the same notebook

````python
import ogma_jupyter as og
````

## 1. Nodes with Data Attributes

Each node can have an `id` and a `data` object containing custom attributes.
These attributes can be used for styling, filtering, or tooltips:

````python
widget1 = og.OgmaWidget()
widget1.graph_data = {
    'nodes': [
        {'id': 'alice', 'x': -80, 'y': -50, 'data': {'label': 'Alice', 'role': 'Engineer', 'department': 'Engineering'}},
        {'id': 'bob', 'x': -80, 'y': 50, 'data': {'label': 'Bob', 'role': 'Manager', 'department': 'Engineering'}},
        {'id': 'carol', 'x': 80, 'y': -50, 'data': {'label': 'Carol', 'role': 'Designer', 'department': 'Design'}},
        {'id': 'dave', 'x': 80, 'y': 50, 'data': {'label': 'Dave', 'role': 'Lead', 'department': 'Design'}}
    ],
    'edges': [
        {'source': 'alice', 'target': 'bob'},
        {'source': 'carol', 'target': 'dave'}
    ]
}
widget1
````

## 2. Edges with Data Attributes

Edges can also have `id` and `data` attributes.
You can use edge attributes to represent relationship types, weights, or other metadata:

````python
widget2 = og.OgmaWidget()
widget2.graph_data = {
    'nodes': [
        {'id': 'server1', 'x': -80, 'y': -60, 'data': {'type': 'web', 'status': 'healthy'}},
        {'id': 'server2', 'x': 80, 'y': -60, 'data': {'type': 'web', 'status': 'healthy'}},
        {'id': 'db', 'x': 0, 'y': 60, 'data': {'type': 'database', 'status': 'healthy'}},
        {'id': 'cache', 'x': 0, 'y': 0, 'data': {'type': 'cache', 'status': 'warning'}}
    ],
    'edges': [
        {'id': 'e1', 'source': 'server1', 'target': 'db', 'data': {'latency_ms': 5, 'type': 'query'}},
        {'id': 'e2', 'source': 'server2', 'target': 'db', 'data': {'latency_ms': 7, 'type': 'query'}},
        {'id': 'e3', 'source': 'server1', 'target': 'cache', 'data': {'latency_ms': 1, 'type': 'read'}},
        {'id': 'e4', 'source': 'server2', 'target': 'cache', 'data': {'latency_ms': 2, 'type': 'read'}}
    ]
}
widget2
````

## 3. Updating Graph Data

You can update the graph data at any time by assigning a new value to `graph_data`.
The widget will automatically re-render with the new data:

````python
# Create widget with initial data
widget3 = og.OgmaWidget()
widget3.graph_data = {
    'nodes': [{'id': 'start', 'x': 0, 'y': 0}],
    'edges': []
}
widget3
````

````python
# Now update it with more data (run this cell to see the change)
widget3.graph_data = {
    'nodes': [
        {'id': 'start', 'x': -100, 'y': 0, 'data': {'label': 'Start'}},
        {'id': 'middle', 'x': 0, 'y': 0, 'data': {'label': 'Middle'}},
        {'id': 'end', 'x': 100, 'y': 0, 'data': {'label': 'End'}}
    ],
    'edges': [
        {'source': 'start', 'target': 'middle'},
        {'source': 'middle', 'target': 'end'}
    ]
}
````

## 4. Multiple Widgets

You can have multiple independent widgets in the same notebook.
Each widget maintains its own graph data and state:

````python
# First widget - simple graph
simple = og.OgmaWidget()
simple.graph_data = {
    'nodes': [{'id': 'a', 'x': -50, 'y': 0}, {'id': 'b', 'x': 50, 'y': 0}],
    'edges': [{'source': 'a', 'target': 'b'}]
}
simple
````

````python
# Second widget - star topology graph
complex_graph = og.OgmaWidget()
complex_graph.graph_data = {
    'nodes': [
        {'id': 'hub', 'x': 0, 'y': 0, 'data': {'label': 'Hub'}},
        {'id': 'spoke1', 'x': 0, 'y': -80, 'data': {'label': 'Spoke 1'}},
        {'id': 'spoke2', 'x': 80, 'y': 0, 'data': {'label': 'Spoke 2'}},
        {'id': 'spoke3', 'x': 0, 'y': 80, 'data': {'label': 'Spoke 3'}},
        {'id': 'spoke4', 'x': -80, 'y': 0, 'data': {'label': 'Spoke 4'}}
    ],
    'edges': [
        {'source': 'hub', 'target': 'spoke1'},
        {'source': 'hub', 'target': 'spoke2'},
        {'source': 'hub', 'target': 'spoke3'},
        {'source': 'hub', 'target': 'spoke4'}
    ]
}
complex_graph
````

## Summary

In this notebook you learned:

1. **Node attributes**: Use the `data` object to store custom properties
2. **Edge attributes**: Edges can have `id` and `data` like nodes
3. **Dynamic updates**: Assign new data to `graph_data` to update the visualization
4. **Multiple widgets**: Each widget is independent and can show different graphs

### Graph Data Structure

```python
graph_data = {
    'nodes': [
        {'id': 'node_id', 'data': {'key': 'value', ...}},
        ...
    ],
    'edges': [
        {'id': 'edge_id', 'source': 'from_id', 'target': 'to_id', 'data': {...}},
        ...
    ]
}
```

- `id` is required for nodes, optional for edges
- `source` and `target` are required for edges
- `data` is optional for both nodes and edges
