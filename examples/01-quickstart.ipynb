# Ogma Jupyter Quickstart

This notebook demonstrates the basic usage of **ogma-jupyter**, a library that brings
Ogma's powerful WebGL graph visualization capabilities to Jupyter notebooks.

With ogma-jupyter, you can:
- Visualize graph data interactively
- Pan and zoom with mouse/trackpad
- Build complex network visualizations entirely in Python

## 1. Import the Library

Import ogma-jupyter using the conventional alias `og`:

````python
import ogma_jupyter as og
````

## 2. Try the Demo

The fastest way to see ogma-jupyter in action is with the built-in demo.
This creates a simple triangle graph with 3 nodes:

````python
og.demo()
````

### What You See

The demo displays:
- **3 nodes** labeled A, B, and C
- **3 edges** connecting them in a triangle pattern

**Try interacting:**
- **Pan**: Click and drag to move the view
- **Zoom**: Use scroll wheel or pinch gesture
- **Click nodes**: Nodes respond to clicks (event handling can be customized)

## 3. Create Your Own Graph

Create a custom graph by setting the `graph_data` property on an `OgmaWidget`:

````python
widget = og.OgmaWidget()
widget.graph_data = {
    'nodes': [{'id': 'a'}, {'id': 'b'}, {'id': 'c'}],
    'edges': [{'source': 'a', 'target': 'b'}, {'source': 'b', 'target': 'c'}]
}
widget
````

This creates a simple chain: `a -> b -> c`
