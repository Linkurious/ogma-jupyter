# ogma-jupyter quickstart

Fast tour of **ogma-jupyter** — interactive graph visualization powered by
[Ogma](https://doc.linkurio.us/ogma/latest/) inside Jupyter.

Covered here:
- Setting your license key
- Displaying a graph from your own data
- Coloring nodes by data properties
- Adding text labels
- Sizing nodes by a numeric field


## 1. License key

Set your Ogma license key once per session.
You can also export it as the `OGMA_LICENSE_KEY` environment variable.


```python
import ogma_jupyter as og

og.set_license("your-license-key")  # or set OGMA_LICENSE_KEY env var
```

```python
og.demo()
```

## 3. Your own graph

Graph data follows Ogma's **RawGraph** schema.
Node positions go under `attributes`; your domain data goes under `data`:


```python
widget = og.OgmaWidget(
    graph_data={
        "nodes": [
            {"id": "a", "attributes": {"x": -100, "y": 0}, "data": {"label": "A"}},
            {"id": "b", "attributes": {"x": 0,    "y": 0}, "data": {"label": "B"}},
            {"id": "c", "attributes": {"x": 100,  "y": 0}, "data": {"label": "C"}},
        ],
        "edges": [
            {"source": "a", "target": "b"},
            {"source": "b", "target": "c"},
        ],
    }
)
widget
```

## 4. Live updates

Assign a new `graph_data` dict at any time — the widget updates in place:


```python
widget.graph_data = {
    "nodes": [
        {"id": "x", "attributes": {"x": -60, "y": 0}, "data": {"label": "X"}},
        {"id": "y", "attributes": {"x": 0,   "y": 0}, "data": {"label": "Y"}},
        {"id": "z", "attributes": {"x": 60,  "y": 0}, "data": {"label": "Z"}},
    ],
    "edges": [
        {"source": "x", "target": "y"},
        {"source": "y", "target": "z"},
        {"source": "z", "target": "x"},
    ],
}
```

## 5. Style rules

Use `add_style_rule()` with helpers from `ogma_jupyter.rules` to bind
visual properties to your data — no manual iteration needed.


```python
from ogma_jupyter import rules

# Rebuild the widget so we start from a clean slate with richer data
styled = og.OgmaWidget(
    graph_data={
        "nodes": [
            {"id": "alice", "attributes": {"x": 0,    "y": -80}, "data": {"name": "Alice", "role": "engineer", "score": 80}},
            {"id": "bob",   "attributes": {"x": -80,  "y": 40},  "data": {"name": "Bob",   "role": "manager",  "score": 55}},
            {"id": "carol", "attributes": {"x": 80,   "y": 40},  "data": {"name": "Carol", "role": "designer", "score": 70}},
        ],
        "edges": [
            {"source": "alice", "target": "bob"},
            {"source": "bob",   "target": "carol"},
            {"source": "carol", "target": "alice"},
        ],
    }
)
styled
```

### Color by category — `rules.map`

Map each discrete value of a data field to an output value.
A `fallback` is used for any value not in the mapping:


```python
styled.add_style_rule(
    node_attributes={
        "color": rules.map(
            field="data.role",
            values={"engineer": "#4e79a7", "manager": "#f28e2b", "designer": "#59a14f"},
            fallback="gray",
        ),
    }
)
```

### Size by number — `rules.slices`

Divide the range of a numeric field into equal slices and map each to an output value.
Here a `score` between 0–100 controls the node radius (6 → 22 px):


```python
styled.add_style_rule(
    node_attributes={
        "radius": rules.slices(
            field="data.score",
            values={"nbSlices": 4, "min": 6, "max": 22},
        ),
    }
)
```

### Text labels — `rules.template`

Build a label string by interpolating `{{field}}` placeholders:


```python
styled.add_style_rule(
    node_attributes={
        "text": {"content": rules.template("{{data.name}}")},
    }
)
```
